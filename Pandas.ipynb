# Pandas: A Python Data Analysis Library

### What is Pandas?

Pandas is one of the most important libraries of Python.

Pandas allows us to analyze big data and make conclusions based on statistical theories. Pandas can clean messy data sets, and make them readable and relevant. Relevant data is very important in data science.

</br>
Pandas has data structures for data analysis. The most used of these are Series and DataFrame data structures. Series is one dimensional, that is, it consists of a column. Data frame is two-dimensional, i.e. it consists of rows and columns.



# Library Highlights

* A fast and efficient **DataFrame** object for data manipulation with integrated indexing;

* Tools for **reading and writing data** between in-memory data structures and different formats: CSV and text files, Microsoft Excel, SQL databases, and the fast HDF5 format;

* Intelligent **data alignment** and integrated handling of **missing data**: gain automatic label-based alignment in computations and easily manipulate messy data into an orderly form;

* Flexible **reshaping** and pivoting of data sets;

* Intelligent label-based **slicing, fancy indexing**, and **subsetting** of large data sets;

* Columns can be inserted and deleted from data structures for **size mutability**;

* Aggregating or transforming data with a powerful **group by** engine allowing split-apply-combine operations on data sets;

* High performance **merging and joining** of data sets;

* **Hierarchical axis indexing** provides an intuitive way of working with high-dimensional data in a lower-dimensional data structure;

* **Time series**-functionality: date range generation and frequency conversion, moving window statistics, date shifting and lagging. Even create domain-specific time offsets and join time series without losing data;

* Highly **optimized for performance**, with critical code paths written in Cython or C.

* Python with pandas is in use in a wide variety of **academic and commercial** domains, including Finance, Neuroscience, Economics, Statistics, Advertising, Web Analytics, and more.


## Real World Applications

* Here are some examples for some real-world fields to illustrate Pandas usage:


    #### 1. Data Science & Machine Learning

    * Example: Handling missing data and feature engineering.

````python

# Sample dataset
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [24, 27, None, 22],
    'Salary': [50000, 54000, 70000, 60000]
})

# Fill missing values and create a new feature
df['Age'] = df['Age'].fillna(df['Age'].mean())  # Fill missing Age with mean value
df['Salary_in_thousands'] = df['Salary'] / 1000  # Create a new feature

print(df)
````

#### 2. Finance & Economics

* Example: Analyzing stock prices using rolling averages.

````python

# Creating a time series of stock prices
date_rng = pd.date_range(start='2023-01-01', end='2023-01-10', freq='D')
stock_prices = pd.DataFrame(date_rng, columns=['Date'])
stock_prices['Price'] = np.random.randint(100, 200, size=(len(date_rng)))

# Set Date as index
stock_prices.set_index('Date', inplace=True)
# Calculate 3-day moving average
stock_prices['3-day MA'] = stock_prices['Price'].rolling(window=3).mean()

print(stock_prices)
````

#### 3. Healthcare & Bioinformatics

* Example: Working with patient data and calculating average values.

````python

# Sample patient data
patient_data = pd.DataFrame({
    'Patient ID': [101, 102, 103, 104],
    'Age': [35, 58, 42, 67],
    'Blood Pressure': [120, 140, 130, 150]
})

# Calculate mean blood pressure for patients over 50 years
mean_bp = patient_data[patient_data['Age'] > 50]['Blood Pressure'].mean()
print(f"Average Blood Pressure for patients over 50: {mean_bp}")
````

#### 4. Marketing & Sales Analytics

* Example: Segmenting customers based on purchase data.

````python

# Sample customer data
customer_data = pd.DataFrame({
    'Customer': ['A', 'B', 'C', 'D'],
    'Purchases': [5, 15, 25, 10]
})

# Create a new column for segmentation based on the number of purchases
customer_data['Segment'] = pd.cut(customer_data['Purchases'], bins=[0, 10, 20, 30], labels=['Low', 'Medium', 'High'])

print(customer_data)
````



### 2. **DataFrame (2D)**
- A **DataFrame** is a two-dimensional data structure, like a table with rows and columns. It is composed of multiple Series objects, each representing a column.
- Think of it as a **spreadsheet** or a **SQL table**.

**Visual Representation:**
```
   Index  Fruit   Color   Quantity
   0      Apple   Red     5
   1      Banana  Yellow  3
   2      Cherry  Red     12
   3      Date    Brown   7
```

In this example:
- Each column (Fruit, Color, Quantity) is a **Series**.
- The entire table is a **DataFrame**.

### Key Differences:
1. **Series**: Single column with an index.
2. **DataFrame**: Multiple columns with an index, where each column is a Series.

## Reshaping & Pivoting in Pandas

````python
data=pd.DataFrame(
    np.arange(16).reshape(4,4),
    index=[list("aabb"),[1,2]*2],
    columns=[["num","num",
              "comp","comp"],
             ["math","stat"]*2])
data
````

````python

long=data.stack()
````

### Pivoting “Long” to “Wide” Format

````python
stock=pd.DataFrame(
    {"fruit": ["apple", "plum","grape"]*2,
     "color": ["purple","yellow"]*3,
     "piece":[3,4,5,6,1,2]})
stock
````

````python
stock.pivot(index="fruit", columns= "color", values= "piece")
````

### Pivoting “Wide” to “Long” Format

````python
data=pd.DataFrame(
    {"lesson":["math","stat","bio"],
     "Sam":[50,60,70],
     "Kim":[80,70,90],
     "Tom":[60,70,85]})
data
````

````python
group=pd.melt(data,["lesson"])
group
````

````python
data=group.pivot(index="lesson", columns="variable", values="value")
data
````

````python
data.reset_index()
````
